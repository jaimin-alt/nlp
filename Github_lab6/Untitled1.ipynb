```python
import random

# ------------------------
# Step 1: Load all sentences from file
# ------------------------
with open("C:/CLG LAB/5TH SEM/NLP/lab5/saved_data/train_set_1lakh.txt", "r", encoding="utf-8") as f:
    all_sentences = [line.strip() for line in f if line.strip()]

print(f"Total sentences loaded: {len(all_sentences)}")  # Should be 98000

# ------------------------
# Step 2: Function to sample 100 sentences and save
# ------------------------
def sample_and_save(all_sentences, method, ngram_type, num_samples=100):
    """
    all_sentences: list of all sentences
    method: 'greedy' or 'beam'
    ngram_type: 'unigram', 'bigram', 'trigram', 'quadrigram'
    num_samples: 100 sentences per file
    """
    sampled = random.sample(all_sentences, num_samples)
    filename = f"{method}_{ngram_type}_sentences.txt"
    with open(filename, "w", encoding="utf-8") as f:
        f.write("\n".join(sampled))
    print(f"Saved {num_samples} sentences to {filename}")

# ------------------------
# Step 3: Generate for all n-grams and both methods
# ------------------------
ngram_types = ['unigram', 'bigram', 'trigram', 'quadrigram']
methods = ['greedy', 'beam']

for method in methods:
    for ngram in ngram_types:
        sample_and_save(all_sentences, method, ngram, num_samples=100)
```
